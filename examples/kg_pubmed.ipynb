```python
# Install required libraries
!pip install pandas py2neo

# Import necessary modules
import pandas as pd
from py2neo import Graph, Node, Relationship

# Set up Neo4j connection
NEO4J_URI = "neo4j+s://04f8ac46.databases.neo4j.io"
NEO4J_USERNAME = "neo4j"
NEO4J_PASSWORD = "d9DCb2j3_4_xAj2-EYHWkNfV9gKrFuU8C_y0yTnGP2w"
graph = Graph(NEO4J_URI, auth=(NEO4J_USERNAME, NEO4J_PASSWORD))

# Read the CSV file
csv_path = '/content/pubmed_articles.csv'
df = pd.read_csv(csv_path)

# Clean the column names
new_columns = ["pmid", "Title", "pmid2", "Abstract", "pmid3", "Author", "pmid4", "Year", "pmid5", "Volume", "pmid6", "Issue", "pmid7", "Journal", "pmid8", "Citation", "pmid9", "Link"]
df.columns = new_columns

# Drop the unnecessary 'pmid' columns
df = df[["pmid", "Title", "Abstract", "Author", "Year", "Volume", "Issue", "Journal", "Citation", "Link"]]

# Function to create or get a node
def get_or_create_node(label, **properties):
    matcher = graph.nodes.match(label, **properties)
    node = matcher.first()
    if node is None:
        node = Node(label, **properties)
        graph.create(node)
    return node

# Iterate over the DataFrame and create nodes and relationships
for index, row in df.iterrows():
    # Create Article node
    article_node = get_or_create_node("Article", pmid=row["pmid"], title=row["Title"], abstract=row["Abstract"])

    # Create Author nodes and relationships
    authors = row["Author"].split(", ")
    for author in authors:
        author_node = get_or_create_node("Author", name=author)
        relationship = Relationship(author_node, "WROTE", article_node)
        graph.create(relationship)

    # Create other nodes (Year, Journal) and relationships if needed
    year_node = get_or_create_node("Year", year=row["Year"])
    graph.create(Relationship(article_node, "PUBLISHED_IN", year_node))

    journal_node = get_or_create_node("Journal", name=row["Journal"])
    graph.create(Relationship(article_node, "PUBLISHED_IN", journal_node))

    # Optionally, handle Volume, Issue, Citation, and Link as needed

print("Knowledge graph created successfully from CSV file.")
```
